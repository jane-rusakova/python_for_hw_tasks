## Завдання 1. Перевірка гіпотези про середній час до покупки (Python)

### 1. Дані та гіпотези

Дано вибірку часу до покупки (в хвилинах) після редизайну сайту:
128, 135, 121, 142, 126, 151, 123

Історичне середнє значення:
μ₀ = 123.7

Нульова гіпотеза:
H₀: μ = 123.7

Альтернативна гіпотеза:
Hₐ: μ > 123.7

Тип тесту: правосторонній one-sample t-test.


```python
import numpy as np
from scipy import stats

data = np.array([128, 135, 121, 142, 126, 151, 123])

# вибіркові статистики
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
sample_size = len(data)

sample_mean, sample_std, sample_size
```

```python
print(f"x̄ = {sample_mean:.2f}")
print(f"s = {sample_std:.2f}")
print(f"n = {sample_size}")
```

Отримані вибіркові статистики:
- x̄ = {sample_mean:.2f}
- s = {sample_std:.2f}
- n = {sample_size}


```python
mu_0 = 123.7
alpha_1 = 0.10
alpha_2 = 0.05

t_stat, p_value = stats.ttest_1samp(data, mu_0, alternative="greater")

t_stat, p_value
```

### 2. Результати t-test

- t-статистика = {t_stat:.2f}
- p-value = {p_value:.4f}

#### Рішення при α = 0.10
Оскільки p-value < 0.10, ми відхиляємо нульову гіпотезу H₀.

#### Рішення при α = 0.05
Оскільки p-value < 0.05, ми також відхиляємо нульову гіпотезу H₀.

### 3. Висновок

На рівнях значущості α = 0.10 та α = 0.05 існують статистично значущі
докази того, що після редизайну сайту середній час до покупки
збільшився.


**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

**Нульова гіпотеза:**
    
    H0:p=0.23H_0: p = 0.23H0:p=0.23
    
**Альтернативна гіпотеза:**
    
    Ha:p>0.23H_a: p > 0.23Ha:p>0.23
    

Тип тесту: **правосторонній z-test для частки**.

```python
from statsmodels.stats.proportion import proportions_ztest, proportion_confint

# дані
count = 311
nobs = 1200
p0 = 0.23
alpha = 0.03

# z-test
z_stat, p_value = proportions_ztest(
    count=count,
    nobs=nobs,
    value=p0,
    alternative='larger'
)

z_stat, p_value
```

```python
print(f"Z-статистика: {z_stat:.4f}")
print(f"p-value: {p_value:.6f}")
print(f"alpha: {alpha}")

if p_value <= alpha:
    print("Рішення: ВІДХИЛЯЄМО H0. Впізнаваність логотипу зросла.")
else:
    print("Рішення: НЕ ВІДХИЛЯЄМО H0. Недостатньо доказів зростання впізнаваності.")
```

### Рішення за результатами z-test

Оскільки p-value < α (0.03), нульову гіпотезу H₀ відхиляємо.
Це означає, що на рівні значущості 3% існують статистично значущі
докази того, що частка дорослих, які впізнають логотип компанії,
стала більшою за 23% після рекламної кампанії.


Оскільки тест односторонній, логічно використовувати

односторонній довірчий інтервал на рівні 1−α=97%1 - \alpha = 97\%1−α=97%.

```python
conf_int = proportion_confint(
    count=count,
    nobs=nobs,
    alpha=alpha,
    method='normal'
)

conf_int
```

97% довірчий інтервал для частки впізнаваності логотипу має вигляд:

p ∈ [L, U]

Нижня межа довірчого інтервалу є більшою за 0.23,
що додатково підтверджує наш висновок з гіпотезного тесту:
фактична частка впізнаваності логотипу після кампанії
статистично значущо перевищує історичний рівень 23%.
